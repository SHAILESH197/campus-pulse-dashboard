```python
import pandas as pd

df = pd.read_csv("../data/student_satisfaction.csv")

df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.histplot(df['satisfaction_score'], bins=5)
plt.title("Satisfaction Score Distribution")
plt.show()
```

```python
df.groupby('facility_rated')['satisfaction_score'].mean().plot(kind='bar')
plt.title("Average Satisfaction by Facility")
plt.show()
```

```python
sns.boxplot(x='academic_year', y='satisfaction_score', data=df)
plt.title("Satisfaction by Academic Year")
plt.show()
```

```python
# Overall average
print("Overall Avg:", df['satisfaction_score'].mean())

# Facility-wise avg
print(df.groupby('facility_rated')['satisfaction_score'].mean())

# Highest rated facility
print("Best Facility:", df.groupby('facility_rated')['satisfaction_score'].mean().idxmax())

# Lowest rated facility
print("Worst Facility:", df.groupby('facility_rated')['satisfaction_score'].mean().idxmin())
```
